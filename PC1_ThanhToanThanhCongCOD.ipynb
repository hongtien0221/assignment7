```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.common.exceptions import NoSuchElementException, NoAlertPresentException
import unittest
import time

class PC03(unittest.TestCase):

    def setUp(self):
        chrome_options = Options()
        chrome_options.add_argument("--headless=new")
        chrome_options.add_argument("--no-sandbox")
        chrome_options.add_argument("--disable-dev-shm-usage")

        self.driver = webdriver.Chrome(service=Service("/usr/bin/chromedriver"), options=chrome_options)

        self.driver.implicitly_wait(30)
        self.base_url = "https://www.google.com/"
        self.verificationErrors = []
        self.accept_next_alert = True

    def test_p_c03(self):
        driver = self.driver
        driver.get("https://nest.botble.com/vi")

        driver.find_element(By.XPATH, "//body[@id='page-home']/header/div[3]/div/div/div[2]/div[2]/div/div[4]/a[2]/span").click()
        driver.find_element(By.LINK_TEXT, "Đăng nhập").click()

        driver.find_element(By.ID, "email").clear()
        driver.find_element(By.ID, "email").send_keys("hien123@gmail.com")
        driver.find_element(By.ID, "password").clear()
        driver.find_element(By.ID, "password").send_keys("hien123@")

        driver.find_element(By.XPATH, "//form[@id='botble-ecommerce-forms-fronts-auth-login-form']/div[4]/button").click()
        driver.find_element(By.LINK_TEXT, "Trang chủ").click()

        # Click vào một sản phẩm
        driver.find_element(By.XPATH, "//main[@id='main-section']/div/div[4]/section/div/div[2]/div[2]/div/div[2]/div[4]/div[2]/a").click()

        # Nhấn vào giỏ hàng
        driver.find_element(By.XPATH, "//img[@alt='Cart']").click()

        # Checkout
        driver.find_element(By.LINK_TEXT, "Proceed To Checkout").click()

        # Coupon
        driver.find_element(By.LINK_TEXT, "You have a coupon code?").click()
        driver.find_element(By.NAME, "coupon_code").clear()
        driver.find_element(By.NAME, "coupon_code").send_keys("SALE10")

        driver.find_element(By.XPATH, "//div[@id='main-checkout-product-info']/div/div[2]/div[4]/div/button").click()

        # Phương thức thanh toán
        driver.find_element(By.XPATH, "//button[@type='submit']").click()

        # Chọn phương thức
        driver.find_element(By.XPATH, "(.//*[normalize-space(text())='Payment method:'])[1]/following::span[1]").click()

        time.sleep(3)

    def tearDown(self):
        self.driver.quit()
        self.assertEqual([], self.verificationErrors)

if __name__ == "__main__":
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
